### Indexing
Items in list and tuples are indexed left to right from 0 or right to left from -1.  
The syntax for indexing is as follows:

```python
data = ['a', 'b', 'c', 'd', 'e']
print(data[0])
print(data[-1])
```

### Slicing
You can select slices of items from list and tuples.  
This is done using the format sequence[start:stop:step].

```python
print(data[0:3:1])  # start at index 0, end at index 3 (which means it will print up to index 2) and print every item.
```

```python
print(data[0::2])  # start at 0, go to the end, print every other item.
```

```python
print(data[:3])  # instead of writing [0:3:1]
```

Note that slicing always produces a shallow copy of the iterable.  
The slice is now a new container.

### Mutability

```python
# a list is mutable
data[0] = 'z'
print(data)
```

```python
# a tuple is immutable
numbers = (1, 2, 3, 4, 5)
numbers[0] = 9
print(numbers)
```

```python
# however if a tuple stores a mutable object
random_tuple = ([1, 2, 3, 4, 5], 'name')
random_tuple[0][0] = 9
print(random_tuple)
```

### Efficiency
tuples are more efficient than list however they lack the mutability of list.
