### What is $\mathcal{O}(1)$?
$\mathcal{O}(1)$, pronounced "Big O of 1", represents constant time complexity.  
It means an algorithm's execution time remains constant and independent of the input size ($n$).

Whether processing 1 item or 1,000,000 items, an $\mathcal{O}(1)$ operation takes roughly the same amount of time and number of steps.

### Common Examples
- Array / List Indexing: Accessing an element by its index (array[5]).
- Hash Map / Dictionary Lookup: Key-value retrieval by key (map["key"]).
- Stack Push/Pop: Adding or removing from the top of a stack.
- Basic Arithmetic: Simple mathematical operations like addition or multiplication (a + b).
- Assignment / Return: Assigning a value to a variable or returning a value.

$O(1)$ does not mean the operation takes 1 millisecond or 1 instruction; it simply means the execution time does not scale with input size.

### $\mathcal{O}(1)$ and Other Complexities
- $\mathcal{O}(1)$ Name = Constant, Scaling with input = flat/constant, Example = reading an item at a known index
- $\mathcal{O}(logn)$ Name = Logarithmic, Scaling = grows very slowly, Example = binary search in a sorted array
- $\mathcal{O}(n)$ Name = Linear, Scaling = grows proportionally with input, Example = loops checking every item once
- $\mathcal{O}(n^2)$ Name = Quadratic, Scaling = grows exponentially with square of input, Example = nested loops over input
