Python comprehensions offer a concise, highly readable way to create collections in a single line.   
They serve as a declarative alternative to standard `for` loops, replacing explicit iteration and append calls with cleaner, faster expressions.

#### The range function
Before we move on there is a function in python used to generate an immutable sequence of numbers.  
It is commonly used in `for` loops to loop a specific number of times.  
It can take up to 3 arguments in the format range(start, stop, step):
- start (which integer to start from) is optional
- stop (which integer to stop at is required)
- step (increments between numbers) is optional

```python
# Example 1
for i in range(5):  # note no start or increment were given and numbering starts from 0 by default
    print(i)
```

```python
# Example 2
for i in range(5, 0, -1):  # we can use a -ve step to count backwards but we must provide a start and stop
    print(i)
```

```python
# Note - attempting to print a range directly only produces what is known as a generator-like range object
print(range(5))
```

```python
# To see the range we can convert it to a list
print(list(range(5)))
```

### List Comprehensions

```python
# Suppose we want the square of all the even numbers from 0 to 10
for x in range(10):
    if x % 2 == 0:
        print(x**2)
```

An easier syntax to achieve this is using a list comprehension which follows the format:
$$\text{Syntax: } [\text{expression} \text{ for } \text{item} \text{ in } \text{iterable} \text{ if } \text{condition}]$$

```python
# So now we can use a list comprehension
squares = [x**2 for x in range(10) if x % 2 == 0]  # this saves our result conveniently to a list for future use
squares
```

```python
# if we just wanted to see a display of the results as with the first 
squares = [print(x**2) for x in range(10) if x % 2 == 0]
```

### Dictionary Comprehensions

Dictionary comprehensions create a mapping of key-value pairs from an iterable using a colon `:` to separate the keys and values.$$\text{Syntax: } \{\text{key\_expr}: \text{val\_expr} \text{ for } \text{item} \text{ in } \text{iterable} \text{ if } \text{condition}\}$$

```python
# Map words to their length, filtering out short words
words = ["python", "data", "list", "comprehension"]
word_lengths = {word: len(word) for word in words if len(word) > 4}
word_lengths
```

### Set Comprehensions

Set comprehensions construct a collection of unique, unordered elements using curly braces `{}` without key-value colons. Duplicate values are automatically deduplicated.$$\text{Syntax: } \{\text{expression} \text{ for } \text{item} \text{ in } \text{iterable} \text{ if } \text{condition}\}$$

```python
# Extract unique word lengths from a list
words = ["apple", "banana", "pear", "apple", "fig"]
unique_lengths = {len(word) for word in words}
unique_lengths
```

### Note
Use IF-ELSE logic: Position `if` at the end for filtering (`[x for x in seq if cond]`).   
Move it before for when using `if-else` transformations (`[x if cond else y for x in seq]`).
