Both Dictionaries and Sets rely on a hash table under the hood, making them essential tools for achieving average $O(1)$ constant time complexity for key operations.

### What are hash tables?

A hash table is an underlying data structure that maps keys to values using a mathematical formula called a hash function.  
It allows looking up, adding, or deleting items nearly instantaneously—in $O(1)$ constant time—regardless of how large the dataset grows.

In Python, both dictionaries (dict) and sets (set) are directly powered by hash maps under the hood.

### How a Hash Map Works Under the Hood

Instead of scanning through an entire list item by item, a hash map converts a key directly into a memory location (an index in an array).
1. Hashing: Python passes a key into its built-in hash() function, which calculates an integer hash value for that key.
2. Bucket Mapping: Python takes that hash value and computes an index (or "bucket") in memory using modular arithmetic (hash % array_size).
3. Direct Lookup: To retrieve or check a key, Python re-computes the hash and jumps directly to that exact memory bucket instead of searching linearly.

```python
# What happens internally during a dict lookup:
key = "apple"

hash_value = hash(key)       # e.g., 592817305847
print(hash_value)

index = hash_value % 8       # e.g., slot 3 in memory array
print(index)

# Python jumps directly to slot 3
```

### How Python Dictionaries and Sets Use Hash Maps

#### Dictionaries:
- Structure = Key-Value pairs
- Hash Table role = maps a key to a specific value
- Primary use case = fast data retrieval by key
- Example = {"age": 30}

#### Sets:
- Structure = Key-only/ unique values
- Hash Table role = store the element itself as the key (no values)
- Primary use case = fast membership testing (item in my_set) and duplicate elimination
- Example = {"apple", "banana"}

### Why Keys Must Be "Hashable" (Immutable)
Because hash maps rely on a stable hash value to pinpoint memory locations, a key's hash value must never change while it is stored in the map.

- Hashable (Allowed as dict keys / set items): Immutable types like int, float, str, tuple, and frozenset. Their contents cannot mutate after creation.
- Unhashable (Not allowed): Mutable types like list, dict, or set. If you modified a list stored as a key, its hash would change, and Python would no longer be able to find where it was saved in memory.

```python
# Valid: Strings and tuples are immutable
my_dict = {("latitude", "longitude"): (37.77, -122.41)}

# Invalid: Raises TypeError: unhashable type: 'list'
my_dict = {[1, 2, 3]: "data"}
```

### Collision Handling in Python
Occasionally, two different keys produce the same hash or evaluate to the same memory slot (a hash collision).

Python handles collisions using a technique called Open Addressing with Quadratic Probing.   
When a slot is already occupied by a different key, Python uses a deterministic algorithm to probe neighboring memory slots until it finds the correct key or an empty slot.

### Set Methods

```python
print(dir(set))
```

A set is an unordered collection of unique hashable items.   
Because sets use hash lookup internally, membership testing (element in set) runs in $O(1)$ time—far faster than checking existence in a list $O(n)$.

```python
# Take the following sets
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}
```

### Primary Set Operations

#### Union ($A \cup B$): Combines elements from both sets, removing duplicates.

```python
set_a | set_b
```

#### Intersection ($A \cap B$): Returns only elements present in both sets.

```python
set_a & set_b
```

#### Difference ($A \setminus B$): Returns elements in set A that are not in set B.

```python
set_a - set_b
```

#### Symmetric Difference ($A \Delta B$): Returns elements present in either set, but not both.

```python
set_a ^ set_b
```
