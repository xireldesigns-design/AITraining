# Nesting

Previously we used the line of code type(print).  
This works in jupyter however if we wrote this code to a .py file and ran it there would be no output.  
We would have to explicitly call print on this line in order to see the output.  
This is where we use nesting as shown below:

```python
print(type(print))  # Notice the output is different to simply using type(print).
```

Now we are seeing some additional info which is that print is an object of the class builtin_function_or_method

Note that while I say nesting functions just remember that type is not a function however it works because type is callable like a function.
