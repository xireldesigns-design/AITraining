# Control Flow (if, elif, else)

### Simple if

Let's say we have an expected outcome called block A.  
Block A is a sequence of lines of code which will be executed as a group.  
Block A will only be executed if a certain condition is met.

```python
# a variable to set up our condition
speed = 100

# the condition
if speed > 80:  # The colon at the end is the token used to initiate block A
    # block A (a block in python is syntactically identified by indentation - typically 4 spaces)
    print("Going too fast")
    slow = speed - 20
    print(slow)

# this code is no longer within block A
print("Halt")
```

In the above example the condition evaluated to True.  
As a result the block of code was executed.  
What would happen if the condition evalauted to False?

```python
speed = 70

# the condition
if speed > 80:
    # block A
    print("Going too fast")
    slow = speed - 20
    print(slow)

print("Halt")
```

block A is not executed!

### Multiple ifs

What about if we have two variables to be used for two conditions.  
The conditions could be used to execute separate blacks of code.  
In this case the outcome of the execution of the second block would be independent of the first.

```python
height = 6  # change 6 to 3 and rerun this cell to see what would happen. Change it back to 6 and rerun to reset the note.
weight = 175

if height > 4:
    print("tall")
if weight > 150:
    print("heavy")
```

### Nested ifs

Or one condition might depend on another.  
In this case if the first condition evalautes to False the second condition will not even be evaluated.

```python
height = 6  # change 6 to 3 and rerun this cell to see what ould happen.  Change it back to 6 and rerun to reset the note.
weight = 175

if height > 4:
    print("tall")
    if weight > 150:
        print("heavy")
```

### Elif
Elifs allow us to navigate multiple outcomes which are related.

```python
price = 100  # Change price to 50 and 150 and rerun the cell the see the different results.

# all of the blocks are based on different conditions related to the price variable
if price == 100:
    print("paid")
elif price < 100:
    print("you still owe ome money")
elif price > 100:
    print("you overpaid")
```

You can nest an if in an if.  
You can also nest an if in elif.  
An elif however can only follow an if and cannot exist on its own.

```python
price = 100  # Change price to 50 and 150 and rerun the cell the see the different results.
response = ""

# all of the blocks are based on different conditions related to the price variable
if price == 100:
    response = "Thank you!"
    print(response)
    if response == response:
        print("Have a good day.")
elif price < 100:
    response = "Wait a minute..."
    print(response)
    if response == response:
        print("You still owe me money.")
elif price > 100:
    response = "Hold up!"
    print(response)
    if response == response:
        print("I have some change for you")
        
```

### Else (Capture any condition related to this variable we may have missed)

```python
temperature = 25  # Change the temperature to 30 or 5 and rerun the cell to see different results

if temperature == 25:
    print("Good!")
elif temperature > 25:
    print("Hot!")
else:
    print("Cold!")
```

### Evaluating Truthiness
Take note that certain values can evaluate to False and are known as Falsy values.  
This i good to know when setting up conditional statements to determine their truthiness.

```python
bool(None)  # None
```

```python
bool(False)  # False
```

```python
bool(0)  # integer 0
```

```python
bool(0.0)  # float 0.0
```

```python
bool("")  # empty string
```

```python
bool([])  # empty list
```

```python
bool(())  # empty set
```

```python
bool({})  # empty dictionary
```

### Ternary (writing a concise if/else)

```python
temp = 25  # change temp to 33 and rerun cell to see the result.
response = "Hot" if temp > 25 else "Cool"
print(response)
```
