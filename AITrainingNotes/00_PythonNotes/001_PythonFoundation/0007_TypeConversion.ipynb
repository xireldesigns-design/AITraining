# Type Conversion (or Type Casting)

There are instances in a program where you might want or need to convert between data types.  
If you look at the list of builtins you can see that the various data types are actually callable.  
They are however not functions but belong to the class type just like type itself.  
This is critical because it means that they are able to create a new instance of the object passed to it of the class it was passed to.  
This happens because type as mentioned before is a powerful class, in fact it is the class that creates all other classes.

```python
print(type(int))
print(type(str))
print(type(list))
```

So if I pass an int to str:

```python
age = 45
print(type(age))
str(age)
print(type(age))
```

You can see even though I passed the int to str the age is still an int.  
However what happened in the line str(age)?

```python
str(age)  # the int age was indeed changed to a s 
```

Remember we would have to reassign the newly created str to the age variable name to store it there:

```python
age = str(age)
print(type(age))
```

So can we change any object to any type? No!  
Lets see what we can and cannot convert aside from int to str.

### To str
You can convert all of the data types we covered to a string e.g.

```python
print(type(str(43.7856)))
print(type(str(3+4j)))
print(type(str(['Paul', 47, 207.8, True, None, 4+9j])))
```

### To int

```python
int(45.8)  # float to int
```

```python
int(True)  # bool to int
```

```python
int('45')  # string with number characters only (0-9) to int. Not even a float can work here.
```

```python
int('34forty-five')  # does not work with strings that contain non number characters
```

None of the container types e.g. list, tuples, sets and dictionaries can be converted to int.

### To Float

```python
float(45)  # int to float
```

```python
float('45.6')  # str with a float number characters or int characters
```

```python
float(True)  # bool to float
```

### To Complex
complex can take two arguments the first being the real part and the second being the imaginary part.  
You separate arguments using a comma as shown below.

```python
complex(4, 6)  # ints to complex
```

```python
complex(2.4, 7.8)   # floats to complex
```

```python
complex(4.4, 2)  # mixture of int and float to complex
```

```python
complex(True, False)  # bools to complex
```

### To List

```python
list(('Paul', 47, 207.8, True, None, 4+9j))  # tuple to a list
```

```python
list({'Paul', 47, 207.8, True, None, 4+9j})  # set to a list
```

```python
list({'Name': 'Paul', 'Age': 47, 'Weight': 207.8, 'IsMale': True})  # dictionary to list only returns the key from the pair
```

In the case of a dictionary to a list you would need to use some methods to get:

```python
archived_userinformation = {'Name': 'Paul', 'Age': 47, 'Weight': 207.8, 'IsMale': True}
print(list(archived_userinformation.values()))  # list of values
print(list(archived_userinformation.items()))  # list of tuples with each key-value [pair]
print(list(archived_userinformation.keys()))  # list of keys explicitly
```

### To Tuple

```python
tuple(['Paul', 47, 207.8, True, None, (4+9j)])  # list to tuple
```

```python
tuple({'Paul', 47, 207.8, True, None, 4+9j})  # set to tuple
```

```python
tuple(archived_userinformation)
```

As with a list you use dictionary methods to extract specific keys, values or key-value pairs

```python
print(tuple(archived_userinformation.values()))  # tuple of values
print(tuple(archived_userinformation.items()))  # tuple of tuples with each key-value [pair]
print(tuple(archived_userinformation.keys()))  # tuple of keys explicitly
```

### To Set

```python
set(['Paul', 47, 207.8, True, None, (4+9j)])  # list to set
```

```python
set(('Paul', 47, 207.8, True, None, 4+9j))  # tuple to set
```

Once again with dictionary methods

```python
set((archived_userinformation))
```

```python
print(set(archived_userinformation.values()))  # set of values
print(set(archived_userinformation.items()))  # set of tuples with each key-value [pair]
print(set(archived_userinformation.keys()))  # set of keys explicitly
```

### To Dictionary

dict() does not directly work as a way to convert types.
