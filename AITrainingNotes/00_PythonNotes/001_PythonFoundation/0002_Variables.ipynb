# Variables

There is a principle in programming called Do not Repeat Yourself or DRY.  
Functions follow the dry principle. As we will learn later functions are actually blocks of prewritten code which we access by their name.  
This way instead of having to type out the entire code of a function everytime we wish to use it we simply call it using its name.  
Similarly there is other code that 1. might be lengthy and or 2. we will use repeatedly in our code.  
To avoid having to write them out every time we can simply create a name called a variable name and assign the object to that name.  
The assignment of that object to a variable name allows us to point the variable name to that object and use it repeatedly in our code.  
Let's see how this works:

```python
greeting = "Hello World!"
```

Let's look at the syntax.
1. We create a variable name and put it to the left of an equal sign.
2. We place our object to the right of that equal sign.

The equal sign (a token) in python is not the same as the equal sign in mathematics, it does not denote equality of value.  
In python the equal sign is called the assignment operator and it allows for the assignment of the object on the right to the name on the left.

Now instead of having to type out "Hello World!" everytime we want to use it, we simply use the variable named greeting:

```python
print(greeting)
```

## Naming Rules for Variables

1. Allowed Characters: Variable names can only contain letters (a-z, A-Z), digits (0-9), and underscores (_).
2. Must Start with a Letter or Underscore: A variable name cannot start with a digit.
3. Case-Sensitive: greeting, Greeting, and GREETING are three completely different variables.
4. No Python Keywords: You cannot use Python’s built-in reserved keywords.
