# Comments

In python the interpreter ignores the hash symbol (#) and anything that comes after it in that line.  
We can use this to make comments in our code as shown below:

```python
# Greet the World!
print("Hello World!")  # The output will be Hello World!
```

It is good practice to make a comment to either preceed or be inline with the code it is refering to.  
It is also good practice to only make comments when necessary and to improve the readability of your code.
