# Methods

Each object type has a set of functions that are specific only to that object type.  
We can see theses class specific functions called methods by looking into the directory of the object type.  
Let's take strings for example:

```python
print(dir(str))
```

The methods start from 'capitalize' and go to 'zfill'.  
Let's take a look at the method lower.  
In order to use the method lower we will learn a new syntax:

```python
# create a string variable
greeting = "HELLO World!"

# use the method lower on the string object
greeting.lower()
```

To use the method first we place the object followed by a . followed by the method followed by open/close parentheses.  
We can see that the output of using lower is to change the entire string to lower case.  
We could check this using help:

```python
help(str.lower)
```

We can check the type of the method:

```python
print(type(str.lower))
```

Note that when we actually call the method we use the parentheses but when we pass it to help or type we leave them off.

```python
# call greeting
print(greeting)
```

Note that the greeting variable still points to "HELLO World!".  
To point greeting to the output of lower we can reassign it as follows:

```python
greeting = greeting.lower()
print(greeting)
```
