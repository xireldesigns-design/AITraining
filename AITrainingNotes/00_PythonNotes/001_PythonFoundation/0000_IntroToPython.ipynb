# Programming

Input - We pass instructions to a computer using a human readable (high level) language.  
Process - These instructions are converted to computer language (low level) for processing via software (compiler / interpreter / hybrid).  
Output - The result from the processed input is returned.

## Python Programming

The Python Interpreter is the software used to convert human readable python language to computer language.  
The Python Interpreter is a hybrid design in that it first compiles the code then interprets it line by line to the processor.

## The Syntax

There are several rules and guidelines which govern how python language is structured.  
Failing to follow these rules and guidelines can result in different kinds of errors.  
It is therefore important to learn the syntax of the python language.

## Concepts

Learning the python syntax is not enough. There are also several critical concepts which allow us to understand how python works.

## Our First Line of Code

```python
print("Hello World!")
```

The python language is structured by using a combination of objects and tokens.  
There are different types of objects. Each type is derived from a blueprint called a class.  
Tokens are a set of reserved keywords and symbols which help to structure the syntax however they are not objects.

print in our first line of code is the name of an object.  
The type of the object that the name print refers to is called a function.  
Functions can be identified by their structure, that is, the function name followed by open and closed parentheses.
These parentheses are tokens which allow the interpreter to know that print is callable (is being used as a function).  
By design a function has parameters which can be passed to the function as arguments.  
The passing is done by placing the argument between the parentheses.  
In the case of print we could execute it without passing any argument as shown below:

```python
print()
```

The output is an empty line.  
In our first line of code we passed an argument to print().  
This argument "Hello World!" is also an object.  
The type of object of "Hello World!" is called a string.  
Strings can be identified by their structure as well. Any character(s) which lie between the open and closed quotation marks (tokens) become part of that string.   
It is important that the quotation marks match. You cannot use a single quotation mark with a double quotation mark.

```python
print("Hello World!')
```

Instead of the expected output we received an error message.  
This is because we used the incorrect syntax and so the python interpreter cannot execute this code.  
It is therefore important to pay attention not only to objects but also the tokens for example, what would happen if we forgot to use the closing parenthesis for a function:

```python
print("Hello World!"
```

We get yet another error message instead of the output we wanted.
