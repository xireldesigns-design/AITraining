# Operators

When we spoke about tokens we mentioned keywords and symbols.  
We also saw how we could manipulate some object types using functions and methods.  
There are a group of tokens we also use in our interaction with some objects which we call operators.  
We have already met one operator in the = sign which we used to assign objects to a variable.  
Let us now take a look at some more operators and see what they do.

## Mathematical

### Addition (+)
We can add numbers using the + operator

```python
salary = 2000
bonus = 500
total_earnings = salary + bonus
print(total_earnings)
```

```python
temp_1 = 25.9
temp_2 = 28
total_temp = temp_1 + temp_2
print(total_temp)
```

```python
Z1 = 2+3j
Z2 = 1+5j
total_Z = Z1 + Z2
print(total_Z)
```

You can also add booleans since True has a value of 1 and False a value of 0.

```python
is_male = True
is_tall = True
score = is_male + is_tall
print(score)
```

You can even add strings which is called concatenation

```python
first_name = 'John'
last_name = 'Brown'
full_name = first_name + " " + last_name  # note a space was also added. A space counts as a character.
print(full_name)
```

You can also concatenate lists and tuples

```python
tuesday_numbers = [10, 16, 12]
wednesday_numbers = [12, 14, 12]
weekly_numbers = tuesday_numbers + wednesday_numbers
print(weekly_numbers)
```

### Floor Division (//)

Floor division outputs only the integer part even if the output carries a fractional part.

```python
2 // 3
```

```python
6.67 // 2.1  # the output is still a float but the fractional part is read as 0
```

### Modulus (%)

While floor division returns the whole number part, modulus calculates the remainder,   
after the left over after dividing the dividend by the divisor.

```python
2 % 3
```

```python
6.67 % 2.1
```

## Logical (and, or, not)
Check multiple conditions at once (and, or)  
Reverse a condition (not)

```python
age_Tom == 43 and age_Jane > 50  # Both conditions are true
```

```python
age_Tom != 43 or age_Jane > 50  # Only one condition needs to be true to return a True value
```

```python
not age_Tom == 43  # not reverses the value of the expected outcome
```

## Identity (is, is not)
Check to see if variables are pointing to the same object in memory

```python
a, b = 2, 2
```

```python
a is b  # a is pointing to the same object in memory as b
```

```python
a is not b
```

```python
# we can use the builtin function id to check this
print(id(a))
print(id(b))  # You cab see both a and b are pointing to an int at the same address in memory
```

## Bitwise operators (&, |, ^, ~, <<, >>)
These operate on intergers at the binary level

```python
c, d = 6, 3
```

```python
# The bin function allows us to see the binary code for integers
print(bin(c))  # 110
print(bin(d))  # 011
```

```python
print(c & d)
print(bin(c & d))  # AND sets each bit to 1 if both bits are 1
```

```python
print(c | d)
print(bin(c | d))  # OR sets each bit to 1 if at least 1 bit is 1
```

```python
print(c ^ d)
print(bin(c ^ d))  # XOR sets each bit to 1 if only 1 of the bits is 1
```

```python
print(~ c)
print(bin(~ c  # NOT flips 1's to 0 and 0's to 1
```

```python
print(c << 1)
print(bin(c << 1))  # Left Shift adds 0 to the right
```

```python
print(c >> 1)
print(bin(c >> 1))  # Right Shift adds 0 the left
```
