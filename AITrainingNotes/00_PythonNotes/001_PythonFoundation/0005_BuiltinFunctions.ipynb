# Builtins

When we did type(print) we saw the output was builtin_function_or_method.  
What is meant by builtin and what is a method?  
Let's examine what a builtin is.  
To do this let's meet our first reserved keyword, called import:

```python
import builtins
```

By the way let's explore import and builtins using type:

```python
print(type(import))  # this produces an error which proves import is not an object.
```

```python
print(type(builtins))  # this is the third type of python object we have now discovered.
```

So far it seems a function carries out a spcific task and a string contain character information but what is a module?  
A module is basically a python file that contains code we might need access to.  
The import keyword gives us access to that code.  
Now we have that access what is in this module?  
We can explore the module using a function called dir (short for directory):

```python
print(dir(builtins))  # if we do not nest dir in print we get a vertical list
```

Looking through this list we see the names of many objects.  
These objects can be used in python right away without having to import them (the list itself is not stored hence why we imported it).  
Why is this distinction being made?  
This is because you can also import objects that other programmers have made using modules,  
You can also create your own objects and use them in your programme directly,  
Or you could even create your own modules with your own objects to share with others and use yourself.

We can find out more about an object by using another object called help().  
Once again we have another example of an object that looks like a function but it is not:

```python
print(type(help))  # help is a object of type _sitebuiltins._Helper
```

Let's use help to explore the print() function:

```python
help(print)  # we can call help without having to nest in print even in a .py
```

This is quite a description.  
Understanding it right now would be difficult.  
For now we learn that print, Prints the values passed to it to the stream or sys.stdout,  
Basically it is saying it displays the arguments passed to it to your monitor!

What about strings?
Well remember the type("Hello World!") is str.  
If we look in the builtins list we can actually see 'str'.  
Let's find out more about it:

```python
help(str)
```
