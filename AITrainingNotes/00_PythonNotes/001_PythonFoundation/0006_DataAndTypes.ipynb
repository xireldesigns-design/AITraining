# Data Types

So far we have met a few types of objects such as builtin functions, strings, type, and _sitebuiltins_.Helper.  
Many times we want to manipulate and store data in our programmes.  
In python we can represent data using objects.  
There are several types of data objects builtin to python.

## Integers
Whole numbers with no fractional part or a decimal point.

```python
age = 46
print(type(age))
```

## Floats
Fractional numbers with a decimal point.

```python
temperature = -12.6
print(type(temperature))
```

## Complex Numbers
Numbers with a real and imaginary part.

```python
Z = 2+3j
type(Z)  # Z is the electrical impedance of an alternating current (AC) circuit.
```

## Boolean Values
True and False

One of the builtin objects is called isinstance:

```python
print(type(isinstance))
```

```python
help(isinstance)
```

isinstance can be used to checks the class of an object.  
It takes two arguments. the actual object and the class we are checking.  
The result is either True or False:

```python
print(isinstance(age, int))
```

```python
print(type(isinstance(age, int)))
```

```python
print(isinstance(age, float))
```

```python
print(type(isinstance(age, float)))
```

As you can see from the outputs both True and False are boolean values.

## Strings
A sequence of characters

```python
user_password = "iujdf%^&SDdkn89"
print(type(user_password))
```

## Lists
A structure that can store a mixture of data types.

```python
user_information = ['Paul', 47, 207.8, True, None, 4+9j]  # We can visually identify lists by their square brackets.
print(type(user_information))
```

## Tuples
Another structure that can store a mixture of data types.

```python
new_user_information = ('Paul', 47, 207.8, True, None, 4+9j)  # We can visually identify tuples by their round brackets.
print(type(new_user_information))
```

## Sets
Yet another structure that can store a mixture of data types.

```python
updated_userinformation = {'Paul', 47, 207.8, True, None, 4+9j}  # We can visually identify sets by their curly brackets.
print(type(updated_userinformation))
```

## Dictionaries
A fourth structure that can hold a mixture of data types.

```python
archived_userinformation = {'Name': 'Paul', 'Age': 47, 'Weight': 207.8, 'IsMale': True}  # returns key-value pairs
print(type(archived_userinformation))
```

With respect to lists, tuples, sets and dictionaries we will learn what is the difference between them and when to use them.
