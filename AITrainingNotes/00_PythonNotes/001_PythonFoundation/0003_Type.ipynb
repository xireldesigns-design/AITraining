# What is the Type of a Python Object?

We have spoken about objects being a core part of python programming.  
We also said that there are various types of objects in python which are derived from a blueprint called a class.  
We encountered two types of objects so far: functions and strings which we used in our first line of code.  
We discovered that we can identify a function and a string by their structure: functionname(), "string"  
There is an easier way to determine the type of an object.  
We use type:

```python
type(print)  # note that print is not just a function, it is a builtin function. We will explore these later.
```

```python
type("Hello World!")  # str is the python shortname for string.
```

What is very interesting is that while type() has the structure of a function it is in fact not a function:

```python
type(type)
```

We will explore what type really is much later but for now just note that it is a very powerful part of the python syntax.
