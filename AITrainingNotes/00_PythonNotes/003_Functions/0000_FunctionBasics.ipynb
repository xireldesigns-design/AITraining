## Builtin vs User Functions

So far we have used built-in functions or functions that we may have imported via modules.  
However sometimes there is no function for a task we want to be executed.  
This is where we have to create our own function.
Let's look at the syntax for the create of a function beginning with the `def` keyword:

```python
def greet():
    print("Function Basics")
```

```python
# Now we can call the function:
greet()  # if you remember we can see that this function has been called without passing any argument
```

#### Parameters and Arguments
Sometimes we need to pass data into our function as we have been doing repeatedly so far.  
In order to do this in our user created function we first need to set up parameters.  
This is done by placing the parameter for the data between the parentheses of the function upon creation.  
Let's explore this below:

```python
def greet_user(message):  # Now he have created a parameter for out function
    print(message)
```

```python
# Now in order to call our function we need to pass an argument to fulfil the parameter we set
greet_user("Hello")  # You can try passing different data types to this function
```

### Types of Arguments

#### 1. Positional Arguments
The arguments must be entered in the position of the parameter

```python
def describe_pet(animal_name, pet_type):
    print(f"Your pet {animal_name}, is a {pet_type}")
```

```python
describe_pet("Aeres", "dog")
```

```python
# Look what happens if you enter them in the wrong position
describe_pet("dog", "Aeres")  # This shows you that the position matters
```

#### 2. Keyword Arguments
We can ignore the positional arguments if we use the paramter names in our arguments

```python
describe_pet(pet_type="dog", animal_name="Aeres")  # we get the order we wanted even though we ignored the positions of the parameters
```

#### 3. Default Parameter Values
Here we can hardcode a default value to the parameter.  
We can enter a different value when we call the function if not the default value is used

```python
def speed(distance=100, time=10):
    print(distance / time)
```

```python
speed()  # Since values have been hardcoded even if no arguments are supplied a value is still printed using the defaults
```

```python
speed(40, 2)  # We can overide the defaults
```

#### A Note on what is returned from a function
If I passed the speed function to the print function I would get:

```python
print(speed())  # what is None?
```

None results because the function has no return value.  
Let's recreate our speed function with a return value:

```python
def new_speed(distance, time):
    return distance / time
```

```python
print(new_speed(40, 2))  # Now that a return value has been defined we no longer see None
```

You can also have multiple values returned (They will be returned as a tuple)

```python
def get_min_max(min, max):
    return min, max
```

```python
result = get_min_max(2, 3)
result
```

```python
# You could separate them using multiple variables
min, max = result
print(f"min = {min},\nmax = {max}")  # note that \ starts an escape sequence and in a string \n creates a newline
```

#### 4. Arbitrary Positional Arguments (`*args`)

The `*args` syntax allows a function to accept any number of additional positional arguments.   
Inside the function, `args` is captured as a tuple.

```python
def sum_all(*args):
    return sum(args)  # we are returning the sum of all of the arguments passed to this function

print(sum_all(1, 2, 3, 4))
```

#### 5. Arbitrary Keyword Arguments (`**kwargs`)
The `**kwargs` syntax allows a function to accept any number of additional keyword arguments.   
Inside the function, `kwargs` is captured as a dictionary.

```python
def print_profile(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

print_profile(username="johndoe", role="admin", active=True)
```

### Combining Parameters and Order
When combining all parameter types in a single function definition, Python requires a strict order:
1. Standard positional parameters
2. Default parameters
3. `*args`
4. Keyword-only parameters (parameters defined after `*args`)
5. `**kwargs`

```python
def complex_function(p1, p2, default_param="default", *args, kw_only, **kwargs):
    print(f"Positional: {p1}, {p2}")
    print(f"Default: {default_param}")
    print(f"Extra positional (*args): {args}")
    print(f"Keyword-only: {kw_only}")
    print(f"Extra keywords (**kwargs): {kwargs}")

complex_function(
    "A", "B",                      # p1, p2
    "custom_default",             # default_param
    "extra_pos1", "extra_pos2",   # *args
    kw_only="required_kw",        # kw_only
    extra_kw1=100, extra_kw2=200  # **kwargs
)
```

### Enforcing Argument Modes (`/` and `*`)
Python provides special operators to explicitly restrict how arguments can be passed:
- `/` (Positional-Only): Parameters before `/` must be passed positionally.
- `*` (Keyword-Only): Parameters after `*` must be passed as keyword arguments.

```python
def strict_function(pos_only, /, standard, *, kw_only):
    print(pos_only, standard, kw_only)

# Valid call:
strict_function("must be pos", "can be either", kw_only="must be kw")
```

```python
# Invalid calls 1:
strict_function(pos_only="error", "standard", kw_only="val")
```

```python
# Invalid call 2:
strict_function("pos", "standard", "error")
```

### Variable Scope
Variables declared inside a function have local scope and exist only within that function execution. Variables declared outside have global scope.

```python
x = 10  # Global scope

def check_scope():
    x = 5  # Local scope (does not alter the global x)
    print("Inside function:", x)

check_scope()
print("Outside function:", x)
```
