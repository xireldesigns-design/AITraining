In traditional programming, data structures like lists load all elements into memory at once (eager evaluation).   
Iterators and Generators enable lazy evaluation—producing items one at a time on demand—which drastically reduces memory footprint when handling massive datasets or infinite streams.

## Core Concepts

### 1. The Iteration Protocol (Iterables & Iterators)
- Iterable: Any Python object capable of returning its members one at a time (e.g., `list`, `tuple`, `dict`, `str`). It implements the `__iter__()` method, which returns an Iterator.
- Iterator: An stateful object that remembers its location during iteration. It implements two primary methods:
  - `__iter__()`: Returns the iterator object itself.
  - `__next__()`: Fetches the next item. When no items remain, it raises a `StopIteration` exception.

```python
# Behind the scenes of a standard 'for' loop
data = [10, 20, 30]
iterator = iter(data)  # Calls data.__iter__()

print(next(iterator))  # Calls iterator.__next__() -> 10
print(next(iterator))  # -> 20
print(next(iterator))  # -> 30
# next(iterator) again would raise StopIteration
```

### 2. Generators & the yield Keyword
A Generator is a simplified way to create an iterator using standard function syntax.  
Instead of returning a final value with `return`, a generator function uses `yield`:
- When `yield` is reached, the function's execution is paused, and the value is sent back to the caller.
- Local variables and execution state are preserved in memory.
- When `next()` is called again, execution resumes immediately after the `yield` statement.

```python
def count_up_to(max_val):
    count = 1
    while count <= max_val:
        yield count  # Pauses execution and returns current count
        count += 1

# Creating generator object (no code inside function runs yet)
counter = count_up_to(3)

print(next(counter))  # Outputs 1
print(next(counter))  # Outputs 2
print(next(counter))  # Outputs 3
```

### Memory Comparison: Eager vs. Lazy Evaluation
When processing large datasets (e.g., 10 million numbers or multi-gigabyte log files), eager evaluation allocates massive blocks of RAM, whereas lazy evaluation maintains a near-zero constant RAM footprint ($\mathcal{O}(1)$ space complexity).

### Practical Applications

#### Reading Massive Files (Chunking)
Loading an 8GB log file with `file.readlines()` will crash or slow down systems with limited memory.  
Using a generator reads line-by-line without loading the whole file into RAM:

```python
def stream_large_file(file_path):
    with open(file_path, "r") as file:
        for line in file:
            yield line.strip()

# Memory usage stays minimal regardless of log file size
for log in stream_large_file("huge_server_logs.txt"):
    if "ERROR" in log:
        process_error(log)
```

#### Generator Expressions
Similar to list comprehensions, generator expressions provide a compact syntax using parentheses `()` instead of brackets `[]`:

```python
# List Comprehension: Creates 10 million items in RAM instantly (~80MB)
list_comp = [x ** 2 for x in range(10_000_000)]

# Generator Expression: Allocates a tiny generator object (~200 bytes)
gen_exp = (x ** 2 for x in range(10_000_000))
```
